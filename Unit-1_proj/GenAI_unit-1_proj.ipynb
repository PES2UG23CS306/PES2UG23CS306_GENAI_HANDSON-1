**LANGUAGE TRSLATION ASSISTANT**

```python
from transformers import pipeline
```

Loading Translation pipelines

English->French:

```python
translator_fr = pipeline(
    "translation_en_to_fr",
    model="Helsinki-NLP/opus-mt-en-fr"
)
```

English->Spanish

```python
translator_es = pipeline(
    "translation_en_to_es",
    model="Helsinki-NLP/opus-mt-en-es"
)
```



Translating text

```python
text = """
This device should be connected to a power source before use.
Ensure that the power button is pressed for three seconds.
"""
```

```python
#Translating to french
french_output = translator_fr(text, max_length=400)
print("French Translation:\n", french_output[0]['translation_text'])
```

```python
#Translating to spanish
spanish_output = translator_es(text, max_length=400)
print("\nSpanish Translation:\n", spanish_output[0]['translation_text'])
```

Reading user manual as input document

```python
def read_manual(file_path):
    with open(file_path, "r", encoding="utf-8") as f:
        return f.read()

manual_text = read_manual("input_eng.txt")
print(manual_text)
```

```python
def translate_manual(text, translator):
    result = translator(text, max_length=500)
    return result[0]["translation_text"]
```

```python
french_manual = translate_manual(manual_text, translator_fr)
print("French Translation:\n", french_manual)
```

```python
spanish_manual = translate_manual(manual_text, translator_es)
print("\nSpanish Translation:\n", spanish_manual)
```

Writing the translated text

```python
with open("manual_french.txt", "w", encoding="utf-8") as f:
    f.write(french_manual)

with open("manual_spanish.txt", "w", encoding="utf-8") as f:
    f.write(spanish_manual)
```
