```python
prompt = "The future of Artificial Intelligence is"
```

EXPERIMEMT-1

```python
models = {
    "BERT": "bert-base-uncased",
    "RoBERTa": "roberta-base",
    "BART": "facebook/bart-base"
}

for name, model in models.items():
    print(f"\n{name} Output:")
    try:
        generator = pipeline("text-generation", model=model)
        output = generator(prompt, max_length=30)
        print(output[0]["generated_text"])
    except Exception as e:
        print("Error:", e)
```

EXPERIMENT-2

```python
from transformers import pipeline

sentence = "The goal of Generative AI is to [MASK] new content."

bert_fill = pipeline("fill-mask", model="bert-base-uncased")
bert_results = bert_fill(sentence)

print("BERT Output:")
for r in bert_results[:3]:
    print(r["token_str"], "| score:", round(r["score"], 4))
```

```python
from transformers import pipeline

sentence = "The goal of Generative AI is to <mask> new content."

roberta_fill = pipeline("fill-mask", model="roberta-base")
roberta_results = roberta_fill(sentence)

print("RoBERTa Output:")
for r in roberta_results[:3]:
    print(r["token_str"], "| score:", round(r["score"], 4))
```

```python
from transformers import pipeline

sentence = "The goal of Generative AI is to <mask> new content."

bart_fill = pipeline("fill-mask", model="facebook/bart-base")
bart_results = bart_fill(sentence)

print("BART Output:")
for r in bart_results[:3]:
    print(r["token_str"], "| score:", round(r["score"], 4))
```

EXPERIMENT-3

```python
context = "Generative AI poses significant risks such as hallucinations, bias, and deepfakes."
question = "What are the risks?"

for name, model in models.items():
    print(f"\n{name} Output:")
    try:
        qa = pipeline("question-answering", model=model)
        answer = qa(question=question, context=context)
        print(answer)
    except Exception as e:
        print("Error:", e)
```
